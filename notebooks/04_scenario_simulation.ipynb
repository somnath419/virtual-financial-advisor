```
# ============================================================
# NOTEBOOK 04 – Scenario Simulation (What-If Analysis)
# Explore savings, expense-cut, and investment projections
# ============================================================
import sys, os

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from dotenv import load_dotenv
load_dotenv(os.path.join(PROJECT_ROOT, '.env'))

print('✅ Project root:', PROJECT_ROOT)
```

```
# ----------------------------------------------------------
# Cell 1 – Load data
# ----------------------------------------------------------
import pandas as pd
from IPython.display import display

DATA_PATH = os.path.join(PROJECT_ROOT, 'data', 'virtual_financial_advisor_data.csv')
df = pd.read_csv(DATA_PATH, parse_dates=['date'])
print(f'Loaded {len(df):,} rows')

USER_ID = 'user_1'
```

```
# ----------------------------------------------------------
# Cell 2 – Scenario: Save 10% more for 12 months
# ----------------------------------------------------------
from src.scenario_simulation import simulate_savings_increase

result = simulate_savings_increase(df, user_id=USER_ID, increase_pct=0.10, months=12)
print(f'=== Scenario: Save 10% More for 12 Months ({USER_ID}) ===')
display(result)
```

```
# ----------------------------------------------------------
# Cell 3 – Scenario: Cut expenses by 15% for 12 months
# ----------------------------------------------------------
from src.scenario_simulation import simulate_expense_reduction

result = simulate_expense_reduction(df, user_id=USER_ID, reduction_pct=0.15, months=12)
print(f'=== Scenario: Cut Expenses by 15% for 12 Months ({USER_ID}) ===')
display(result)
```

```
# ----------------------------------------------------------
# Cell 4 – Scenario: Invest savings at 7% p.a. for 10 years
# ----------------------------------------------------------
from src.scenario_simulation import simulate_investment_growth

result = simulate_investment_growth(df, user_id=USER_ID, annual_return_rate=0.07, years=10)
print(f'=== Scenario: Invest at 7% for 10 Years ({USER_ID}) ===')
display(result)
```

```
# ----------------------------------------------------------
# Cell 5 – Custom scenarios (experiment with parameters)
# ----------------------------------------------------------
print('=== Custom: Save 20% More for 24 Months ===')
display(simulate_savings_increase(df, user_id=USER_ID, increase_pct=0.20, months=24))

print('\n=== Custom: Cut Expenses by 25% for 6 Months ===')
display(simulate_expense_reduction(df, user_id=USER_ID, reduction_pct=0.25, months=6))

print('\n=== Custom: Invest at 10% p.a. for 20 Years ===')
display(simulate_investment_growth(df, user_id=USER_ID, annual_return_rate=0.10, years=20))
```

```
# ----------------------------------------------------------
# Cell 6 – Compare savings increase scenarios across users
# ----------------------------------------------------------
all_users = sorted(df['user_id'].unique())
rows = []
for uid in all_users:
    s = simulate_savings_increase(df, uid, increase_pct=0.10, months=12)
    e = simulate_expense_reduction(df, uid, reduction_pct=0.15, months=12)
    i = simulate_investment_growth(df, uid, annual_return_rate=0.07, years=10)
    rows.append({
        'user_id':            uid,
        'extra_savings_10pct': s['extra_savings'],
        'saved_by_cut_15pct':  e['total_savings_gained'],
        'portfolio_10yr_7pct': i['projected_portfolio_value'],
    })

scenarios_df = pd.DataFrame(rows).set_index('user_id')
print('=== Scenario Comparison – All Users ===')
display(scenarios_df)
```
