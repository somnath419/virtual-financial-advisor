```python
# ============================================================
# NOTEBOOK 01 – Data Loading & Exploratory Data Analysis
# ============================================================
import sys, os

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from dotenv import load_dotenv
load_dotenv(os.path.join(PROJECT_ROOT, '.env'))

print('✅ Project root:', PROJECT_ROOT)
```

```python
# ----------------------------------------------------------
# Cell 1 – Load transaction data
# ----------------------------------------------------------
import pandas as pd
from IPython.display import display

DATA_PATH = os.path.join(PROJECT_ROOT, 'data', 'virtual_financial_advisor_data.csv')
df = pd.read_csv(DATA_PATH, parse_dates=['date'])

print(f'Loaded {len(df):,} rows  |  {df["user_id"].nunique()} unique users')
display(df.head())
```

```python
# ----------------------------------------------------------
# Cell 2 – Data shape & types
# ----------------------------------------------------------
print('Shape:', df.shape)
print('\nColumn dtypes:')
print(df.dtypes)
```

```python
# ----------------------------------------------------------
# Cell 3 – Descriptive statistics
# ----------------------------------------------------------
print('--- Descriptive Statistics ---')
display(df.describe())
```

```python
# ----------------------------------------------------------
# Cell 4 – Category & user distributions
# ----------------------------------------------------------
print('--- Category Value Counts ---')
display(df['category'].value_counts())

print('\n--- User Transaction Counts ---')
display(df['user_id'].value_counts())
```

```python
# ----------------------------------------------------------
# Cell 5 – Missing values
# ----------------------------------------------------------
print('--- Missing Values ---')
display(df.isnull().sum())
```

```python
# ----------------------------------------------------------
# Cell 6 – Date range
# ----------------------------------------------------------
print(f'Date range : {df["date"].min().date()}  →  {df["date"].max().date()}')
print(f'Months covered : {df["date"].dt.to_period("M").nunique()}')
```

```python
# ----------------------------------------------------------
# Cell 7 – Income vs Expense split (all users)
# ----------------------------------------------------------
income_total  = df[df['amount'] > 0]['amount'].sum()
expense_total = df[df['amount'] < 0]['amount'].abs().sum()
print(f'Total Income   : ${income_total:,.2f}')
print(f'Total Expenses : ${expense_total:,.2f}')
print(f'Net            : ${income_total - expense_total:,.2f}')
```
