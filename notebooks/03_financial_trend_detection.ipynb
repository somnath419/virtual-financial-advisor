```
# ============================================================
# NOTEBOOK 03 – Financial Trend Detection
# Health metrics + risky pattern analysis
# ============================================================
import sys, os

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from dotenv import load_dotenv
load_dotenv(os.path.join(PROJECT_ROOT, '.env'))

print('✅ Project root:', PROJECT_ROOT)
```

```
# ----------------------------------------------------------
# Cell 1 – Load data
# ----------------------------------------------------------
import pandas as pd
from IPython.display import display

DATA_PATH = os.path.join(PROJECT_ROOT, 'data', 'virtual_financial_advisor_data.csv')
df = pd.read_csv(DATA_PATH, parse_dates=['date'])
print(f'Loaded {len(df):,} rows')
```

```
# ----------------------------------------------------------
# Cell 2 – Financial health metrics for a single user
# ----------------------------------------------------------
from src.trend_detection import financial_health_metrics

USER_ID = 'user_1'

metrics = financial_health_metrics(df, user_id=USER_ID)
print(f'=== Financial Health Metrics for {USER_ID} ===')
display(metrics)
```

```
# ----------------------------------------------------------
# Cell 3 – Detect risky patterns for a single user
# ----------------------------------------------------------
from src.trend_detection import detect_risky_patterns

risks = detect_risky_patterns(df, user_id=USER_ID)
print(f'=== Risky Patterns for {USER_ID} ({len(risks)} found) ===')
if risks:
    for i, risk in enumerate(risks, 1):
        print(f"\n[{i}] Type     : {risk['type']}")
        print(f"    Severity : {risk['severity']}")
        print(f"    Message  : {risk['message']}")
        print(f"    Details  : {risk['details']}")
else:
    print('No risky patterns detected.')
```

```
# ----------------------------------------------------------
# Cell 4 – Health metrics for ALL users (summary table)
# ----------------------------------------------------------
all_users = sorted(df['user_id'].unique())
rows = []
for uid in all_users:
    m = financial_health_metrics(df, uid)
    rows.append({
        'user_id':        uid,
        'total_income':   m['total_income'],
        'total_expenses': m['total_expenses'],
        'net_savings':    round(m['total_income'] - m['total_expenses'], 2),
        'savings_rate':   m['savings_rate'],
    })

summary_df = pd.DataFrame(rows).set_index('user_id')
print('=== Health Metrics Summary – All Users ===')
display(summary_df)
```

```
# ----------------------------------------------------------
# Cell 5 – Risky pattern summary for ALL users
# ----------------------------------------------------------
risk_rows = []
for uid in all_users:
    user_risks = detect_risky_patterns(df, uid)
    for r in user_risks:
        risk_rows.append({'user_id': uid, **r})

if risk_rows:
    risk_df = pd.DataFrame(risk_rows)[['user_id', 'type', 'severity', 'message']]
    print('=== All Risky Patterns Across All Users ===')
    display(risk_df)
else:
    print('No risky patterns detected across any user.')
```
