```
# ============================================================
# NOTEBOOK 02 – Expense Classification
# Classify transactions as Income / Expense / Neutral
# ============================================================
import sys, os

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from dotenv import load_dotenv
load_dotenv(os.path.join(PROJECT_ROOT, '.env'))

# ----------------------------------------------------------
# Cell 1 – Load data
# ----------------------------------------------------------
import pandas as pd
from IPython.display import display

DATA_PATH = os.path.join(PROJECT_ROOT, 'data', 'virtual_financial_advisor_data.csv')
df = pd.read_csv(DATA_PATH, parse_dates=['date'])

# ----------------------------------------------------------
# Cell 2 – Classify each transaction
# ----------------------------------------------------------
from src.expense_classifier import classify_transaction

df['type'] = df['amount'].apply(classify_transaction)

print('--- Transaction type distribution ---')
display(df['type'].value_counts())

# ----------------------------------------------------------
# Cell 3 – Total amount by type
# ----------------------------------------------------------
print('--- Total amount by type ---')
display(df.groupby('type')['amount'].sum().rename('total_amount'))
```
